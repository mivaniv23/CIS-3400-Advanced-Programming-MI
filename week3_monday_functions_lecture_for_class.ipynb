## 1. `return` vs. `print` — the big one

These two functions look almost identical. They are not.

````python
def add_and_print(a, b):
    print(a + b)


def add_and_return(a, b):
    return a + b
````

**Predict before you run:** what will the next cell display? Write your guess down
before running it.

````python
x = add_and_print(3, 4)
y = add_and_return(3, 4)

print("x is:", x)
print("y is:", y)
````

`add_and_print` put `7` on the screen and then handed back **nothing**. In Python,
"nothing" is a real value called `None`.

`add_and_return` put nothing on the screen and handed back `7`, which we stored in `y`.

> A function with no `return` statement always returns `None`.

### Why this matters

A printed value is gone. You cannot do anything else with it.

````python
# This will fail. Read the error message carefully before we discuss it.
total = add_and_print(3, 4) * 2
````

`TypeError: unsupported operand type(s) for *: 'NoneType' and 'int'`

Translated: *you tried to multiply nothing by 2.*

Now the version that works:

````python
total = add_and_return(3, 4) * 2
print(total)
````

### The rule to remember

**`print` is for humans. `return` is for the rest of your program.**

A function that only prints is a dead end; nothing else in your code can use the result.
When you are unsure, `return` the value and let the *caller* decide whether to print it.

> **Notebook warning:** if you put a function call on the *last line* of a cell, Jupyter
> displays the returned value automatically. That is a notebook convenience, not the
> function printing. Assign the result to a variable — like we did above — when you want
> to see what is really happening.

## 2. Parameters, arguments, and defaults

- A **parameter** is the name in the `def` line.
- An **argument** is the actual value you pass in when you call it.

Notice the """ - we will discuss this a little later!

````python
def make_label(name, department, active=True):
    """Build a display label for an employee record."""
    status = "active" if active else "inactive"
    return f"{name} ({department}) - {status}"


print(make_label("Rivera", "Finance"))
````

`active=True` is a **default value**. If the caller does not supply it, Python uses
the default. Parameters with defaults must come *after* parameters without them.

### Positional vs. keyword arguments

````python
# Positional: matched by order
print(make_label("Rivera", "Finance"))

# Keyword: matched by name, so order stops mattering
print(make_label(department="Finance", name="Rivera"))

# Mixed: positional arguments must come first
print(make_label("Rivera", department="Finance", active=False))
````

**Predict before you run:** what does this next cell do? Does it crash?

````python
print(make_label("Finance", "Rivera"))
````

It runs. It just produces a wrong answer!

This is worth pausing on. A crash tells you where the problem is. A silent wrong answer
does not, and it can survive all the way into a report someone makes a decision from.

**Practical habit:** once a function takes more than two or three parameters, call it with
keyword arguments. It costs you a few keystrokes and buys you a bug you never have to find.

## 3. Scope — where a variable lives

Run this and read the error.

````python
def compute_total(prices):
    running = 0
    for p in prices:
        running += p
    return running


print(compute_total([10, 20, 30]))
print(running)
````

`running` was created *inside* the function. When the function finished, it was
thrown away. It does not exist out here.

That is not a limitation, it is why you can use `i` or `total`
inside a function without worrying about what someone else named their variables.


**Predict the following before you run.**

````python
count = 10


def bump():
    count = 99
    return count


print("bump() returns:", bump())
print("count is still:", count)
````

Assigning to `count` inside the function created a brand-new *local* `count`. The
outer one was never touched.

### Reading from outside is allowed

A function *can* see variables defined outside of it, it just cannot reassign them without
extra work.

````python
TAX_RATE = 0.07


def with_tax(amount):
    return amount * (1 + TAX_RATE)


print(with_tax(100))
````

This is fine for a constant like a tax rate. It is a bad idea for anything that
changes, because now your function's answer depends on invisible state somewhere else in
the file.

**Preferred:** pass what you need in as a parameter. A function whose output depends only
on its inputs is one you can test, reuse, and reason about.

## 4. Docstrings and type hints

Two habits we are adopting for the rest of the course.

````python
def with_tax(amount: float, rate: float = 0.07) -> float:
    """Return the amount with sales tax applied.

    Args:
        amount: The pre-tax amount in dollars.
        rate: The tax rate as a decimal. Defaults to 0.07.

    Returns:
        The total including tax.
    """
    return amount * (1 + rate)


print(with_tax(100))
print(with_tax(100, rate=0.05))
````

### Which part is the docstring?

All of it: every line from the opening """ down to the closing """, blank lines
and all. It is one string that happens to span nine lines.

```
def with_tax(amount: float, rate: float = 0.07) -> float:
    """Return the amount with sales tax applied.        <-- docstring STARTS here

    Args:
        amount: The pre-tax amount in dollars.
        rate: The tax rate as a decimal. Defaults to 0.07.

    Returns:
        The total including tax.
    """                                                 <-- and ENDS here
    return amount * (1 + rate)                     <-- back to ordinary code
```

Two things make it a docstring rather than just a string sitting in your function:

1. **It comes first.** The docstring must be the very first statement in the body, before
   any code. Put it lower down and Python builds the string, throws it away, and your
   function ends up with no documentation at all.
2. **Python keeps it.** It gets attached to the function as `with_tax.__doc__` — which is
   exactly what `help()` reads back, and what VS Code shows you when you hover over a
   function name.

The triple quotes are not magic. They simply let a string run across multiple lines, which
is the only reason we need them here. A one-line docstring in ordinary quotes works fine
too, but the convention is triple quotes even for one line, so there is room to grow.

**The shape we will use in this course** (this format is called Google style):

- One line saying what the function returns, in plain English
- A blank line
- `Args:` — one line per parameter, noting the default if it has one
- `Returns:` — what comes back

You will meet other formats, especially "NumPy" style. Most of them are fine in
industry. In our class, use this one.

The type hints (`amount: float`, `-> float`) do not change how the code runs. Python
will not stop you from passing a string. What they do is tell the next reader — usually
future you — what the function expects and what it gives back. VS Code reads them too, and
will autocomplete and warn you accordingly.

The docstring is what `help()` shows:

````python
help(with_tax)
````

> **Course standard from here on:** every function you write should get a docstring and type
> hints. It makes next week's object-oriented work far
> easier to read.
